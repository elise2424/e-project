```python
import numpy as np
import matplotlib.pyplot as plt

# ===== 1) Material constants (from your doc) =====
E = 70e9           # Young's Modulus [Pa]  (70 GPa)
sigma_y = 150e6    # Yield stress [Pa]     (150 MPa)
K = 400e6          # Strength coefficient [Pa] (400 MPa)
n = 0.2            # Strain hardening exponent [-]

# ===== 2) Yield strain inferred from Hooke's law =====
#    σ_y = E * ε_y  ->  ε_y = σ_y / E
eps_y = sigma_y / E   # ~0.00214 for the given values

# ===== 3) Strain range for plotting (adjust as needed) =====
eps_max = 0.20        # up to 20% strain for demonstration
N = 1000
eps = np.linspace(0.0, eps_max, N)

# ===== 4) Piecewise stress definition =====
# Elastic region (ε <= ε_y): σ = E * ε
sigma_elastic = E * eps

# Plastic region (ε > ε_y): σ = σ_y + K * (ε - ε_y)^n
# Use np.maximum to avoid negative inside power
plastic_part = sigma_y + K * np.maximum(eps - eps_y, 0.0) ** n

# Combine piecewise curve
sigma = np.where(eps <= eps_y, sigma_elastic, plastic_part)

# ===== 5) Plot (Stress in MPa) =====
plt.figure(figsize=(7, 5))
plt.plot(eps, sigma/1e6, linewidth=2)
plt.axvline(eps_y, color='gray', linestyle='--', linewidth=1, label=f'Yield strain ≈ {eps_y:.4f}')
plt.title('Stress–Strain Curve (Aluminium Alloy)\nElastic + Hollomon Plastic')
plt.xlabel('Strain ε (–)')
plt.ylabel('Stress σ (MPa)')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# ===== 6) Print key points for discussion =====
uts_idx = np.argmax(sigma)  # for this simple model, stress keeps rising with ε
print("=== Key Points ===")
print(f"Young's Modulus E: {E/1e9:.1f} GPa")
print(f"Yield stress σ_y:   {sigma_y/1e6:.1f} MPa")
print(f"Yield strain ε_y:   {eps_y:.5f} (≈ {eps_y*100:.2f}%)")
print(f"Max plotted strain: {eps_max:.2f} (≈ {eps_max*100:.0f}%)")
```
